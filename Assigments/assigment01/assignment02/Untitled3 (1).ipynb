```python
# ==============================
# Student Performance Analytics System
# ==============================

# ---------- Dataset ----------
students_data = [
    {"id": "S01", "name": "Ali", "scores": (78, 85, 90)},
    {"id": "S02", "name": "Sara", "scores": (88, 92, 79)},
    {"id": "S03", "name": "Ahmed", "scores": (45, 60, 55)},
    {"id": "S04", "name": "Zain", "scores": (95, 91, 89)}
]

# ==============================
# Part A – Functional Design
# ==============================

# 1️⃣ Calculate Average
def calculate_average(scores):
    return sum(scores) / len(scores)


# 2️⃣ Determine Grade
def determine_grade(avg):
    if avg >= 85:
        return "A"
    elif 70 <= avg <= 84:
        return "B"
    elif 50 <= avg <= 69:
        return "C"
    else:
        return "Fail"


# 3️⃣ Identify Top Student
def get_top_student(data):
    top_name = None
    highest_avg = 0

    for student in data:
        avg = calculate_average(student["scores"])
        if avg > highest_avg:
            highest_avg = avg
            top_name = student["name"]

    return top_name


# 4️⃣ Students Who Passed All Subjects
def get_passed_students(data):
    passed_students = []

    for student in data:
        if all(score >= 50 for score in student["scores"]):
            passed_students.append(student["name"])

    return passed_students


# ==============================
# Part B – Object-Oriented Design
# ==============================

class StudentAnalytics:

    def __init__(self, students_data):
        self.students_data = students_data

    # Adds average & grade to each student
    def compute_results(self):
        for student in self.students_data:
            avg = calculate_average(student["scores"])
            grade = determine_grade(avg)
            student["average"] = avg
            student["grade"] = grade

    # Get top student
    def get_top_student(self):
        return max(self.students_data, key=lambda s: s["average"])["name"]

    # Get class average
    def get_class_average(self):
        total = sum(student["average"] for student in self.students_data)
        return total / len(self.students_data)

    # Get unique grades (set)
    def get_unique_grades(self):
        return {student["grade"] for student in self.students_data}

    # Generate structured report
    def generate_report(self):
        grade_distribution = {}

        for student in self.students_data:
            grade = student["grade"]
            grade_distribution[grade] = grade_distribution.get(grade, 0) + 1

        return {
            "class_average": self.get_class_average(),
            "top_student": self.get_top_student(),
            "grade_distribution": grade_distribution
        }

    # Advanced: Consistently Improving Students
    def get_consistent_improvers(self):
        improving_students = []

        for student in self.students_data:
            s = student["scores"]
            if s[0] < s[1] < s[2]:
                improving_students.append(student["name"])

        return improving_students

    # Advanced: Convert to Tabular Structure
    def to_tabular(self):
        return {
            "ids": [s["id"] for s in self.students_data],
            "names": [s["name"] for s in self.students_data],
            "averages": [s["average"] for s in self.students_data],
            "grades": [s["grade"] for s in self.students_data]
        }


# ==============================
# Example Execution
# ==============================

if __name__ == "__main__":

    # Functional Results
    print("Top Student (Functional):", get_top_student(students_data))
    print("Passed Students:", get_passed_students(students_data))

    # OOP Results
    analytics = StudentAnalytics(students_data)
    analytics.compute_results()

    print("Top Student (OOP):", analytics.get_top_student())
    print("Class Average:", analytics.get_class_average())
    print("Unique Grades:", analytics.get_unique_grades())
    print("Report:", analytics.generate_report())
    print("Consistent Improvers:", analytics.get_consistent_improvers())
    print("Tabular Format:", analytics.to_tabular())
```

```python
# ==========================================
# Course Enrollment & Performance System
# ==========================================

# ---------- Dataset ----------
courses_data = {
    "Python": {
        "instructor": "Dr. A",
        "students": {"S01", "S02", "S03"}
    },
    "Machine Learning": {
        "instructor": "Dr. B",
        "students": {"S02", "S04"}
    },
    "Data Science": {
        "instructor": "Dr. C",
        "students": {"S01", "S04"}
    }
}

# ==========================================
# Part A – Functional Requirements
# ==========================================

# 1️⃣ Students enrolled in more than one course
def get_multi_course_students(courses):
    student_count = {}

    for course in courses.values():
        for student in course["students"]:
            student_count[student] = student_count.get(student, 0) + 1

    return [student for student, count in student_count.items() if count > 1]


# 2️⃣ Courses with more than 2 students
def get_large_courses(courses):
    return [name for name, info in courses.items()
            if len(info["students"]) > 2]


# 3️⃣ Student → Number of Enrolled Courses
def get_student_course_count(courses):
    student_count = {}

    for course in courses.values():
        for student in course["students"]:
            student_count[student] = student_count.get(student, 0) + 1

    return student_count


# 4️⃣ All Unique Students
def get_all_unique_students(courses):
    unique_students = set()

    for course in courses.values():
        unique_students.update(course["students"])

    return unique_students


# ==========================================
# Part B – Object-Oriented Design
# ==========================================

class CourseAnalytics:

    def __init__(self, courses_data):
        self.courses_data = courses_data

    def get_multi_course_students(self):
        return get_multi_course_students(self.courses_data)

    def get_student_course_count(self):
        return get_student_course_count(self.courses_data)

    def get_largest_course(self):
        return max(self.courses_data.items(),
                   key=lambda item: len(item[1]["students"]))[0]

    def generate_course_report(self):
        return {
            "total_courses": len(self.courses_data),
            "total_unique_students": len(get_all_unique_students(self.courses_data)),
            "largest_course": self.get_largest_course(),
            "student_course_distribution": self.get_student_course_count()
        }

    # ==========================================
    # Advanced Requirement
    # Convert course-centered → student-centered
    # ==========================================
    def to_student_centered(self):
        student_courses = {}

        for course_name, info in self.courses_data.items():
            for student in info["students"]:
                student_courses.setdefault(student, []).append(course_name)

        return student_courses


# ==========================================
# Example Execution
# ==========================================

if __name__ == "__main__":

    # Functional Results
    print("Students in Multiple Courses:", get_multi_course_students(courses_data))
    print("Courses with >2 Students:", get_large_courses(courses_data))
    print("Student Course Count:", get_student_course_count(courses_data))
    print("All Unique Students:", get_all_unique_students(courses_data))

    # OOP Results
    analytics = CourseAnalytics(courses_data)

    print("Largest Course:", analytics.get_largest_course())
    print("Course Report:", analytics.generate_course_report())
    print("Student-Centered Structure:", analytics.to_student_centered())
```
