```python
import numpy as np

sensor_data = np.array([
    [12, 15, 14, 10, 13],
    [20, 22, 19, 18, 21],
    [30, 28, 27, 29, 31],
    [25, 24, 26, 23, 22]
])

avg_per_machine = np.mean(sensor_data, axis=1)
print("Average per machine:", avg_per_machine)

max_per_column = np.max(sensor_data, axis=0)
print("Max per time interval:", max_per_column)

normalized_data = sensor_data / max_per_column
print("Normalized Matrix:\n", normalized_data)

machines_above_20 = np.where(avg_per_machine > 20)[0]
print("Machines with avg > 20 (index):", machines_above_20)

reshaped_matrix = sensor_data.reshape(2, 10)
print("Reshaped Matrix (2x10):\n", reshaped_matrix)
```

```python
import numpy as np

A = np.random.randint(1, 51, (3, 3))
B = np.random.randint(1, 51, (3, 3))

print("Matrix A:\n", A)
print("Matrix B:\n", B)

matrix_mult = A @ B
print("Matrix Multiplication (A @ B):\n", matrix_mult)

elementwise_mult = A * B
print("Element-wise Multiplication:\n", elementwise_mult)

greater_than_100 = matrix_mult[matrix_mult > 100]
print("Elements > 100:\n", greater_than_100)

A[A < 10] = 0
B[B < 10] = 0

print("Matrix A after replacing <10:\n", A)
print("Matrix B after replacing <10:\n", B)

stacked_matrix = np.vstack((A, B))
print("Vertically Stacked Matrix:\n", stacked_matrix)
```
